Import dependencies

```python
import numpy as np
from digit_classify import digit_classify
#NOTE: When testing with your own data, you don't have to import sys, os, load_data libraries and add your current working directory to the PATH. 
import sys
import os
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), '..'))) # You need this for utils folder functions
from utils.dataloader import load_data # Using training data to test the function
```

```python
# Load data
X_test, y_test = load_data('training_data')
```

```python
N = len(X_test)
predictions = []

# Predicting
for i in range(N):
    X = X_test[i]
    pred = digit_classify(X)
    predictions.append(pred)
```

Checking accuracy

```python
predictions = np.array(predictions)
y_test = np.array(y_test)
accuracy = np.mean(predictions == y_test)
print(f'The accuracy of the MLR model is {accuracy:.4f}.')
```
