# Feature selection/extraction/engineering: (29.10)

I believe this to be the most crucial step, and I am not too sure what we should implement here... Since the data is in some sense time series data, perhaps there are some cycles and common frequencies (fourier transform?)

```python
# Idea 1: take the length between the first and the last 3D point (i can already assume "0" and "8" to not be well-separated)
# Idea 2: means of each axis (x,y,z), probably not a good one 
dfTotal = pd.DataFrame()
for j in range(len(doubleList)): # placeholder loop for feature engineering from the initial data samples
    eng_feat_1 = []
    eng_feat_2 = []
    for i in range(len(doubleList[j])):
        sample_i = doubleList[j][i].to_numpy() # originally every matrix is stored as a pandas df, but since the same feature
        # extraction should be done on the test matrices provided by the teacher, it is probably a good idea to make sure the logic works like this
        firstRow = sample_i[0,:]
        lastRow = sample_i[-1,:]
        distance = np.sqrt(((firstRow - lastRow)**2).sum(axis=0)) # Euclidian distance. Now every sample has 1 feature
    
        eng_feat_1.append(distance)
        eng_feat_2.append(sample_i.mean(axis = 0)) # another set of features (means of each axis)

    feat2 = np.stack(eng_feat_2, axis = 0) # stacking into a matrix

    df = pd.DataFrame(eng_feat_1) # first column feature (Euclidian distance between 1st and last points)
    df[1] = feat2[:,0] # this probably does not follow "best practises"
    df[2] = feat2[:,1]
    df[3] = feat2[:,2]
    df[4] = j # column labels, note: keep this as the last column such that the logic does not break further down
    dfTotal = pd.concat([dfTotal,df], axis = 0) # concatenating every sample to a large dataframe for each class
# now "dfTotal" has set of feature columns (X), and a label column (y)
#print(dfTotal)

avgs = []
for i in range(10):
    mean_i = dfTotal.loc[dfTotal.iloc[:,-1] == i, 0].mean()
    print("Number drawn:",i,",average length:",mean_i)
    avgs.append(mean_i)

nums = np.linspace(0,9,10)
fig, ax = plt.subplots()
ax.scatter(nums,avgs)
ax.set_xticks(nums)
ax.set_xlabel("first/last point distances, averages of each class (0-9)")
plt.show()

# test/train split (validation set)
from sklearn.model_selection import train_test_split

X = dfTotal.iloc[:,:-1]
y = dfTotal.iloc[:,-1]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2) # random shuffle

# using a premade model just to get an idea of the current features
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns

model = KNeighborsClassifier(n_neighbors=5)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

CR_RF = classification_report(y_test, y_pred)
CM_RF = confusion_matrix(y_test, y_pred)
print(CR_RF)
sns.heatmap(CM_RF, annot=True, cmap = "coolwarm")
plt.show()
# ~40% accuracy over all,  better than the minimum requirement (>10%),
# I have a suitable kNN code and I doubt there would be any performance issues.
# Although these features are not good
```
